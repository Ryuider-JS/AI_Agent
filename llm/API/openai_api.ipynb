**3. 텍스트 생성 (Generations)**

GPT 기반 모델을 사용하여 질문에 대한 답변을 생성할 수 있음

**파라미터 설명**

* model: 사용할 모델 (예: "gpt-4o-mini")
* messages: 대화의 흐름을 나타내는 메시지 리스트
  * role: 메시지 작성자의 역할 ("system", "user" 중 하나)
  * content: 메시지의 실제 내용
* max_tokens: 생성할 응답의 최대 토큰 수 (옵션, 기본값 있음)
* temperature: 응답의 창의성을 조절하는 값 (0에서 1 사이, 낮을수록 고정적인 답변)
* n: 생성할 응답의 개수 (기본값: 1)


````python
from openai import OpenAI

response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is a LLM?"}
  ]
)
````

````python
# 응답에서 메시지 내용 추출 및 출력
answer = response.choices[0].message.content
print(f"생성된 답변: \n{answer}")
````

## 이미지

- 이미지 생성
- 이미지 변형

**1. 라이브러리 불러오기**

````python
from openai import OpenAI
````

**2. OpenAI API 키 설정**

````python
api_key = ""
client = OpenAI(api_key=api_key)
````

**3. 이미지 생성 (Generations)**

DALL-E 3 모델을 사용할 경우 이미지 크기는 1024x1024, 1024x1792, 1792x1024 픽셀 중 선택 가능

**파라미터 설명**

* model: 사용할 모델 (현재 "dall-e-3" 또는 "dall-e-2" 지원)
* prompt: 이미지를 생성하기 위한 텍스트 설명
* size: 생성할 이미지의 크기 (예: "1024x1024")
* quality: 이미지 품질 (DALL-E 3 에서만 사용 가능, "standard" 또는 "hd" 선택)
* n: 생성할 이미지 개수 (DALL-E 3 에서는 1개만 생성 가능, DALL-E 2 에서는 최대 10개 생성)


````python

# DALL-E 3을 사용하여 하얀 샴 고양이 이미지 생성
response = client.images.generate(
  model="dall-e-3",
  prompt="a white siamese cat",
  size="1024x1024",
  quality="standard",
  n=1,
)

# 이미지 URL 추출
image_url = response.data[0].url

# 이미지 URL 출력
print(f"생성된 이미지 URL: {image_url}")
````

**DALL-E 3 특징**

* DALL-E 3 모델은 안전상의 이유와 더욱 상세한 이미지 생성을 위해 입력 프롬프트를 자동으로 수정
* 내가 하고자 하는 프롬프트를 명확하게 반영하기 위해서는 프롬프트 앞에 다음과 같은 문구를 추가해야 함

```
I NEED to test how the tool works with extremely simple prompts. DO NOT add any detail, just use it AS-IS:
```


**4. 이미지 입력을 통한 변형 (Variations - DALL-E 2 only)**

* 이미지 변형 기능은 기존 이미지를 기반으로 새로운 변형 이미지를 생성
* 주어진 이미지의 스타일이나 구성을 유지하면서 새로운 이미지를 만들어낼 수 있음

**파라미터 설명**

* model: 사용할 모델 ("dall-e-2"만 지원)
* image: 변형할 이미지 (바이트 데이터 또는 파일 경로)
* n: 생성할 이미지 개수 (최대 10개)
* size: 생성할 이미지의 크기 (예: "1024x1024")

````python
# 이미지 파일 경로 설정
image_path = "./data/corgi_and_cat_paw.png"

# 이미지 변형
response = client.images.create_variation(
  model="dall-e-2",
  image=open(image_path, "rb"),
  n=1,
  size="1024x1024"
)

# 이미지 URL 추출
image_url = response.data[0].url

# 이미지 URL 출력
print(f"생성된 이미지 URL: {image_url}")
````

**5. 이미지 입력을 통한 질문**

* 이미지 입력을 통해 모델에게 원하는 질문을 할 수 있음
* 주어진 이미지의 스타일이나 구성을 유지하면서 새로운 이미지를 만들어낼 수 있음

**파라미터 설명**

* model: 사용할 모델 ("gpt-4o, gpt-4o-mini, gpt-4-turbo" 지원)
* image_url: 변형할 이미지 (웹상에 업로드되어 있는 url 리스트)

````python
response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[
    {
      "role": "user",
      "content": [
        {"type": "text", "text": "이 이미지 안에 뭐가 그려져 있어?"},
        {
          "type": "image_url",
          "image_url": {
            "url": "https://upload.wikimedia.org/wikipedia/commons/thumb/d/dd/Gfp-wisconsin-madison-the-nature-boardwalk.jpg/2560px-Gfp-wisconsin-madison-the-nature-boardwalk.jpg",
          },
        },
      ],
    }
  ],
  max_tokens=300,
)
answer = response.choices[0]

print(f"생성된 답변: {answer.message.content}")
````

````python
import base64

# 로컬 이미지 파일 경로
image_path = "./data/corgi_and_cat_paw.png"

# 이미지 파일을 base64로 인코딩하는 함수
def encode_image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")
    
# base64로 인코딩된 이미지 데이터
encoded_image = encode_image_to_base64(image_path)

response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[
    {
      "role": "user",
      "content": [
        {"type": "text", "text": "이 이미지 안에 뭐가 그려져 있어?"},
        {
          "type": "image_url",
          "image_url": {
            "url": f"data:image/jpeg;base64,{encoded_image}"
          }
        }
      ],
    }
  ],
  max_tokens=300,
)

answer = response.choices[0]

print(f"생성된 답변: {answer.message.content}")
````
