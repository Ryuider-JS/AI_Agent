### 2. 활용 기초 실습

#### 2.1. 기본 텍스트 생성 실습


```python
import google.generativeai as genai

# API 키 설정
genai.configure(api_key=os.environ['GOOGLE_API_KEY'])

# 모델 초기화
model = genai.GenerativeModel('gemini-2.5-flash')

# 프롬프트 전달
response = model.generate_content("인공지능의 미래에 대해 설명해주세요.")
print(response.text)
```

#### 2.2. 이미지 분석 실습

```python
!pip install pillow
```

```python
import PIL.Image

# 이미지 파일 로드
image = PIL.Image.open('./data/corgi_and_cat_paw.png')

# 모델 초기화
model = genai.GenerativeModel('gemini-2.5-pro')

# 프롬프트 전달
response = model.generate_content(["이 이미지에 대해 설명해주세요.", image])
print(response.text)
```

#### 2.3. 대화형 응답 생성 실습

```python
# 채팅 세션 시작
chat = model.start_chat(history=[])

# 대화 진행
response = chat.send_message("인공지능의 장단점은 무엇인가요?")
print(response.text)

response = chat.send_message("그렇다면 인공지능의 윤리적 문제는 어떻게 해결할 수 있을까요?")
print(response.text)
```
